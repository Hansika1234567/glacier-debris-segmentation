```python
def prithvi_features(model, x):

    """
    Extract multi-scale Prithvi features
    from transformer blocks:

    7
    15
    23
    31
    """

    # Input:
    # [B, C, H, W]
    #
    # Prithvi patch embedding expects:
    # [B, C, T, H, W]
    #
    # We have one temporal frame.

    if (
        len(x.shape) == 4
        and
        model.patch_embed.input_size[0] == 1
    ):

        x = x.unsqueeze(2)

    sample_shape = x.shape[-3:]

    # Patch embedding
    x = model.patch_embed(x)

    # Positional embedding
    pos_embed = (
        model.interpolate_pos_encoding(
            sample_shape
        )
    )

    x = x + pos_embed[:, 1:, :]

    # No temporal/location encoding
    # No random masking for segmentation

    # CLS token
    cls_token = (
        model.cls_token
        +
        pos_embed[:, :1, :]
    )

    cls_tokens = cls_token.expand(
        x.shape[0],
        -1,
        -1
    )

    x = torch.cat(
        (cls_tokens, x),
        dim=1
    )

    features = []

    # Transformer blocks
    for i, block in enumerate(
        model.blocks
    ):

        x = block(x)

        if i in [
            7,
            15,
            23,
            31
        ]:

            features.append(x)

    # Final normalization
    x = model.norm(x)

    return features
```

```python
def tokens_to_feature_maps(features):

    spatial_features = []

    for f in features:

        # Remove CLS token
        f = f[:, 1:, :]

        B, N, C = f.shape

        H = W = int(
            N ** 0.5
        )

        # [B, N, C]
        # ↓
        # [B, C, N]

        f = f.transpose(
            1,
            2
        )

        # ↓
        # [B, C, H, W]

        f = f.reshape(
            B,
            C,
            H,
            W
        )

        spatial_features.append(
            f
        )

    return spatial_features
```

```python
import torch.nn as nn
import torch.nn.functional as F


def make_norm(channels):

    # Stable for batch size = 1

    groups = (
        32
        if channels >= 32
        else 1
    )

    return nn.GroupNorm(
        groups,
        channels
    )
```

```python
class ConvModule(nn.Module):

    def __init__(
        self,
        in_channels,
        out_channels,
        kernel_size=3
    ):

        super().__init__()

        padding = (
            kernel_size // 2
        )

        self.block = nn.Sequential(

            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size,
                padding=padding,
                bias=False
            ),

            make_norm(
                out_channels
            ),

            nn.ReLU(
                inplace=True
            )
        )


    def forward(self, x):

        return self.block(x)
```

```python
class PyramidPoolingModule(nn.Module):

    def __init__(
        self,
        in_channels,
        out_channels,
        pool_scales=(
            1,
            2,
            3,
            6
        )
    ):

        super().__init__()

        self.pool_scales = (
            pool_scales
        )

        self.convs = (
            nn.ModuleList([

                nn.Sequential(

                    nn.Conv2d(
                        in_channels,
                        out_channels,
                        kernel_size=1,
                        bias=False
                    ),

                    make_norm(
                        out_channels
                    ),

                    nn.ReLU(
                        inplace=True
                    )

                )

                for _ in pool_scales

            ])
        )


        self.bottleneck = (
            ConvModule(

                in_channels
                +
                len(pool_scales)
                *
                out_channels,

                out_channels,

                kernel_size=3
            )
        )


    def forward(self, x):

        outputs = [x]


        for scale, conv in zip(

            self.pool_scales,

            self.convs

        ):

            pooled = (
                F.adaptive_avg_pool2d(

                    x,

                    output_size=(
                        scale,
                        scale
                    )
                )
            )


            pooled = conv(
                pooled
            )


            pooled = (
                F.interpolate(

                    pooled,

                    size=x.shape[-2:],

                    mode="bilinear",

                    align_corners=False

                )
            )


            outputs.append(
                pooled
            )


        x = torch.cat(

            outputs,

            dim=1

        )


        return self.bottleneck(
            x
        )
```

```python
class UPerNetDecoder(nn.Module):

    def __init__(

        self,

        in_channels=1280,

        decoder_channels=256,

        num_classes=2

    ):

        super().__init__()


        # PPM

        self.ppm = (
            PyramidPoolingModule(

                in_channels,

                decoder_channels

            )
        )


        # Lateral convolutions

        self.lateral_convs = (
            nn.ModuleList([

                ConvModule(

                    in_channels,

                    decoder_channels,

                    kernel_size=1

                )

                for _ in range(3)

            ])
        )


        # FPN convolutions

        self.fpn_convs = (
            nn.ModuleList([

                ConvModule(

                    decoder_channels,

                    decoder_channels,

                    kernel_size=3

                )

                for _ in range(4)

            ])
        )


        # FPN bottleneck

        self.fpn_bottleneck = (
            ConvModule(

                decoder_channels * 4,

                decoder_channels,

                kernel_size=3

            )
        )


        # Segmentation head

        self.classifier = (
            nn.Sequential(

                nn.Conv2d(

                    decoder_channels,

                    128,

                    kernel_size=3,

                    padding=1

                ),

                make_norm(
                    128
                ),

                nn.ReLU(
                    inplace=True
                ),


                nn.Conv2d(

                    128,

                    64,

                    kernel_size=3,

                    padding=1

                ),

                make_norm(
                    64
                ),

                nn.ReLU(
                    inplace=True
                ),


                nn.Dropout2d(
                    0.1
                ),


                nn.Conv2d(

                    64,

                    num_classes,

                    kernel_size=1

                )

            )
        )


    def forward(
        self,
        features
    ):


        c1, c2, c3, c4 = (
            features
        )


        # PPM only on deepest feature

        p4 = self.ppm(
            c4
        )


        # Lateral features

        l1 = (
            self.lateral_convs[0](
                c1
            )
        )


        l2 = (
            self.lateral_convs[1](
                c2
            )
        )


        l3 = (
            self.lateral_convs[2](
                c3
            )
        )


        # Top-down FPN


        l3 = l3 + (
            F.interpolate(

                p4,

                size=l3.shape[-2:],

                mode="bilinear",

                align_corners=False

            )
        )


        l2 = l2 + (
            F.interpolate(

                l3,

                size=l2.shape[-2:],

                mode="bilinear",

                align_corners=False

            )
        )


        l1 = l1 + (
            F.interpolate(

                l2,

                size=l1.shape[-2:],

                mode="bilinear",

                align_corners=False

            )
        )


        # FPN refinement


        f1 = self.fpn_convs[0](
            l1
        )

        f2 = self.fpn_convs[1](
            l2
        )

        f3 = self.fpn_convs[2](
            l3
        )

        f4 = self.fpn_convs[3](
            p4
        )


        target_size = (
            f1.shape[-2:]
        )


        f2 = F.interpolate(

            f2,

            size=target_size,

            mode="bilinear",

            align_corners=False

        )


        f3 = F.interpolate(

            f3,

            size=target_size,

            mode="bilinear",

            align_corners=False

        )


        f4 = F.interpolate(

            f4,

            size=target_size,

            mode="bilinear",

            align_corners=False

        )


        # Concatenate all levels


        x = torch.cat(

            [
                f1,
                f2,
                f3,
                f4
            ],

            dim=1

        )


        x = self.fpn_bottleneck(
            x
        )


        return self.classifier(
            x
        )
```

```python
class PrithviUPerNet(nn.Module):

    def __init__(
        self,
        encoder,
        decoder
    ):

        super().__init__()

        self.encoder = encoder

        self.decoder = decoder


    def forward(
        self,
        x
    ):


        features = (
            prithvi_features(

                self.encoder,

                x

            )
        )


        features = (
            tokens_to_feature_maps(

                features

            )
        )


        logits = self.decoder(

            features

        )


        logits = F.interpolate(

            logits,

            size=x.shape[-2:],

            mode="bilinear",

            align_corners=False

        )


        return logits
```

```python
import os
import glob
import numpy as np
import torch
import rasterio

from torch.utils.data import Dataset, DataLoader
import torch.nn.functional as F


class GSSDDataset(Dataset):

    def __init__(
        self,
        root="/content/GSSD/GSSD",
        split="train",
        augment=False
    ):

        # Image directory
        self.image_dir = os.path.join(
            root,
            split,
            "image"
        )

        # Mask directory
        self.mask_dir = os.path.join(
            root,
            split,
            "mask"
        )

        self.augment = augment


        # Get all image TIFF files
        self.image_files = sorted(

            glob.glob(
                os.path.join(
                    self.image_dir,
                    "image*.tif"
                )
            ),

            key=lambda x: int(
                os.path.basename(x)
                .replace("image", "")
                .replace(".tif", "")
            )
        )


        # Create corresponding mask paths
        self.mask_files = [

            os.path.join(
                self.mask_dir,

                "mask" +

                os.path.basename(x)
                .replace("image", "")
            )

            for x in self.image_files
        ]


        # Check that every image has a mask
        for img, mask in zip(
            self.image_files,
            self.mask_files
        ):

            if not os.path.exists(mask):

                raise FileNotFoundError(
                    f"Missing mask for:\n"
                    f"Image: {img}\n"
                    f"Expected mask: {mask}"
                )


        print(
            f"{split.upper()} dataset loaded: "
            f"{len(self.image_files)} samples"
        )


    def __len__(self):

        return len(self.image_files)


    def __getitem__(self, idx):


        # --------------------------------
        # Read first 6 spectral bands
        # --------------------------------

        with rasterio.open(
            self.image_files[idx]
        ) as src:

            image = src.read(
                [
                    1,  # Blue
                    2,  # Green
                    3,  # Red
                    4,  # NIR
                    5,  # SWIR1
                    6   # SWIR2
                ]
            ).astype(np.float32)


        # --------------------------------
        # Read ground-truth mask
        # --------------------------------

        with rasterio.open(
            self.mask_files[idx]
        ) as src:

            mask = src.read(
                1
            ).astype(np.float32)


        # --------------------------------
        # Convert NumPy → PyTorch tensors
        # --------------------------------

        image = torch.from_numpy(
            image
        )

        mask = torch.from_numpy(
            mask
        )


        # --------------------------------
        # Resize image
        #
        # [6, 512, 512]
        #       ↓
        # [6, 448, 448]
        # --------------------------------

        image = F.interpolate(

            image.unsqueeze(0),

            size=(448, 448),

            mode="bilinear",

            align_corners=False

        ).squeeze(0)


        # --------------------------------
        # Resize mask
        #
        # [512, 512]
        #       ↓ unsqueeze
        # [1, 1, 512, 512]
        #       ↓ resize
        # [1, 1, 448, 448]
        #       ↓ squeeze
        # [448, 448]
        # --------------------------------

        mask = F.interpolate(

            mask
            .unsqueeze(0)
            .unsqueeze(0),

            size=(448, 448),

            mode="nearest"

        )

        mask = mask.squeeze(0).squeeze(0)


        # --------------------------------
        # Data augmentation
        # Training only
        # --------------------------------

        if self.augment:


            # Horizontal flip
            if torch.rand(1).item() < 0.5:

                image = torch.flip(
                    image,
                    dims=[2]
                )

                mask = torch.flip(
                    mask,
                    dims=[1]
                )


            # Vertical flip
            if torch.rand(1).item() < 0.5:

                image = torch.flip(
                    image,
                    dims=[1]
                )

                mask = torch.flip(
                    mask,
                    dims=[0]
                )


        # --------------------------------
        # Convert mask to binary classes
        # --------------------------------

        mask = (
            mask > 0.5
        ).long()


        return (
            image,
            mask
        )
```

```python
class FocalLoss(
    nn.Module
):


    def __init__(

        self,

        alpha=0.25,

        gamma=2.0

    ):

        super().__init__()


        self.alpha = (
            alpha
        )


        self.gamma = (
            gamma
        )


    def forward(

        self,

        logits,

        targets

    ):


        ce = F.cross_entropy(

            logits,

            targets,

            reduction="none"

        )


        pt = torch.exp(
            -ce
        )


        focal = (

            self.alpha

            *

            (
                1 - pt
            )

            **

            self.gamma

            *

            ce

        )


        return focal.mean()
```

```python
def segmentation_metrics(logits, targets):

    preds = logits.argmax(dim=1)

    pred = preds.flatten()
    target = targets.flatten()

    # Background
    bg_intersection = (
        ((pred == 0) & (target == 0))
        .sum()
        .float()
    )

    bg_union = (
        ((pred == 0) | (target == 0))
        .sum()
        .float()
    )

    # Debris / Glacier
    debris_intersection = (
        ((pred == 1) & (target == 1))
        .sum()
        .float()
    )

    debris_union = (
        ((pred == 1) | (target == 1))
        .sum()
        .float()
    )

    debris_pred = (
        (pred == 1)
        .sum()
        .float()
    )

    debris_true = (
        (target == 1)
        .sum()
        .float()
    )

    precision = (
        debris_intersection /
        (debris_pred + 1e-7)
    )

    recall = (
        debris_intersection /
        (debris_true + 1e-7)
    )

    f1 = (
        2 * precision * recall /
        (precision + recall + 1e-7)
    )

    iou_bg = (
        bg_intersection /
        (bg_union + 1e-7)
    )

    iou_debris = (
        debris_intersection /
        (debris_union + 1e-7)
    )

    # CORRECT mean IoU
    miou = (
        iou_bg + iou_debris
    ) / 2.0

    return {
        "iou": iou_debris.item(),
        "miou": miou.item(),
        "precision": precision.item(),
        "recall": recall.item(),
        "f1": f1.item()
    }
```

```python
def train_one_epoch(

    net,

    loader,

    optimizer,

    criterion,

    scaler,

    accumulation_steps=8

):


    net.train()


    running_loss = 0.0


    optimizer.zero_grad(

        set_to_none=True

    )


    for step, (

        images,

        masks

    ) in enumerate(
        loader
    ):


        images = images.to(

            device,

            non_blocking=True

        )


        masks = masks.to(

            device,

            non_blocking=True

        )


        with torch.autocast(

            device_type="cuda",

            dtype=torch.float16

        ):


            logits = net(
                images
            )


            loss = criterion(

                logits,

                masks

            )


            loss = (

                loss

                /

                accumulation_steps

            )


        scaler.scale(
            loss
        ).backward()


        if (

            (
                step + 1
            )

            %

            accumulation_steps

            == 0

            or

            (
                step + 1
            )

            ==

            len(loader)

        ):


            scaler.unscale_(
                optimizer
            )


            torch.nn.utils.clip_grad_norm_(

                net.parameters(),

                max_norm=1.0

            )


            scaler.step(
                optimizer
            )


            scaler.update()


            optimizer.zero_grad(

                set_to_none=True

            )


        running_loss += (

            loss.item()

            *

            accumulation_steps

        )


        if step % 50 == 0:


            print(

                f"Step {step}/{len(loader)} "

                f"Loss: "

                f"{loss.item() * accumulation_steps:.4f}"

            )


    return (

        running_loss

        /

        len(loader)

    )
```

```python
@torch.no_grad()

def validate(

    net,

    loader,

    criterion

):


    net.eval()


    total_loss = 0.0


    metrics = {

        "iou": 0.0,

        "miou": 0.0,

        "precision": 0.0,

        "recall": 0.0,

        "f1": 0.0

    }


    count = 0


    for images, masks in loader:


        images = images.to(

            device,

            non_blocking=True

        )


        masks = masks.to(

            device,

            non_blocking=True

        )


        with torch.autocast(

            device_type="cuda",

            dtype=torch.float16

        ):


            logits = net(
                images
            )


            loss = criterion(

                logits,

                masks

            )


        total_loss += (
            loss.item()
        )


        batch_metrics = (

            segmentation_metrics(

                logits,

                masks

            )

        )


        for key in metrics:

            metrics[key] += (

                batch_metrics[key]

            )


        count += 1


    for key in metrics:

        metrics[key] /= count


    return (

        total_loss / count,

        metrics

    )
```

```python
def prithvi_features(model, x):

    """
    Extract multi-scale Prithvi features
    from transformer blocks:

    7
    15
    23
    31
    """

    # Input:
    # [B, C, H, W]
    #
    # Prithvi patch embedding expects:
    # [B, C, T, H, W]
    #
    # We have one temporal frame.

    if (
        len(x.shape) == 4
        and
        model.patch_embed.input_size[0] == 1
    ):

        x = x.unsqueeze(2)

    sample_shape = x.shape[-3:]

    # Patch embedding

    x = model.patch_embed(x)

    # Positional embedding

    pos_embed = (
        model.interpolate_pos_encoding(
            sample_shape
        )
    )

    x = x + pos_embed[:, 1:, :]

    # No temporal/location encoding
    # No random masking for segmentation

    # CLS token

    cls_token = (
        model.cls_token
        +
        pos_embed[:, :1, :]
    )

    cls_tokens = cls_token.expand(
        x.shape[0],
        -1,
        -1
    )

    x = torch.cat(
        (cls_tokens, x),
        dim=1
    )

    features = []

    # Transformer blocks

    for i, block in enumerate(
        model.blocks
    ):

        x = block(x)

        if i in [
            7,
            15,
            23,
            31
        ]:

            features.append(x)

    # Final normalization

    x = model.norm(x)

    return features
```

```python
def tokens_to_feature_maps(features):

    spatial_features = []

    for f in features:

        # Remove CLS token

        f = f[:, 1:, :]

        B, N, C = f.shape

        H = W = int(
            N ** 0.5
        )

        # [B, N, C]
        # ↓
        # [B, C, N]

        f = f.transpose(
            1,
            2
        )

        # ↓
        # [B, C, H, W]

        f = f.reshape(
            B,
            C,
            H,
            W
        )

        spatial_features.append(
            f
        )

    return spatial_features
```

```python
def make_norm(channels):

    # Stable for batch size = 1

    groups = (
        32
        if channels >= 32
        else 1
    )

    return nn.GroupNorm(
        groups,
        channels
    )
```

```python
class ConvModule(nn.Module):

    def __init__(
        self,
        in_channels,
        out_channels,
        kernel_size=3
    ):

        super().__init__()

        padding = (
            kernel_size // 2
        )

        self.block = nn.Sequential(

            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size,
                padding=padding,
                bias=False
            ),

            make_norm(
                out_channels
            ),

            nn.ReLU(
                inplace=True
            )
        )

    def forward(self, x):

        return self.block(x)
```

```python
class PyramidPoolingModule(nn.Module):

    def __init__(
        self,
        in_channels,
        out_channels,
        pool_scales=(
            1,
            2,
            3,
            6
        )
    ):

        super().__init__()

        self.pool_scales = (
            pool_scales
        )

        self.convs = (
            nn.ModuleList([

                nn.Sequential(

                    nn.Conv2d(
                        in_channels,
                        out_channels,
                        kernel_size=1,
                        bias=False
                    ),

                    make_norm(
                        out_channels
                    ),

                    nn.ReLU(
                        inplace=True
                    )

                )

                for _ in pool_scales

            ])
        )

        self.bottleneck = (
            ConvModule(

                in_channels
                +
                len(pool_scales)
                *
                out_channels,

                out_channels,

                kernel_size=3
            )
        )

    def forward(self, x):

        outputs = [x]

        for scale, conv in zip(

            self.pool_scales,

            self.convs

        ):

            pooled = (
                F.adaptive_avg_pool2d(

                    x,

                    output_size=(

                        scale,
                        scale

                    )
                )
            )

            pooled = conv(
                pooled
            )

            pooled = (
                F.interpolate(

                    pooled,

                    size=x.shape[-2:],

                    mode="bilinear",

                    align_corners=False

                )
            )

            outputs.append(
                pooled
            )

        x = torch.cat(

            outputs,

            dim=1

        )

        return self.bottleneck(
            x
        )
```

```python
class UPerNetDecoder(nn.Module):

    def __init__(

        self,

        in_channels=1280,

        decoder_channels=256,

        num_classes=2

    ):

        super().__init__()

        # PPM

        self.ppm = (

            PyramidPoolingModule(

                in_channels,

                decoder_channels

            )

        )

        # Lateral convolutions

        self.lateral_convs = (

            nn.ModuleList([

                ConvModule(

                    in_channels,

                    decoder_channels,

                    kernel_size=1

                )

                for _ in range(3)

            ])

        )

        # FPN convolutions

        self.fpn_convs = (

            nn.ModuleList([

                ConvModule(

                    decoder_channels,

                    decoder_channels,

                    kernel_size=3

                )

                for _ in range(4)

            ])

        )

        # FPN bottleneck

        self.fpn_bottleneck = (

            ConvModule(

                decoder_channels * 4,

                decoder_channels,

                kernel_size=3

            )

        )

        # Segmentation head

        self.classifier = (

            nn.Sequential(

                nn.Conv2d(

                    decoder_channels,

                    128,

                    kernel_size=3,

                    padding=1

                ),

                make_norm(
                    128
                ),

                nn.ReLU(
                    inplace=True
                ),

                nn.Conv2d(

                    128,

                    64,

                    kernel_size=3,

                    padding=1

                ),

                make_norm(
                    64
                ),

                nn.ReLU(
                    inplace=True
                ),

                nn.Dropout2d(
                    0.1
                ),

                nn.Conv2d(

                    64,

                    num_classes,

                    kernel_size=1

                )

            )

        )

    def forward(

        self,
        features

    ):

        c1, c2, c3, c4 = (
            features
        )

        # PPM only on deepest feature

        p4 = self.ppm(
            c4
        )

        # Lateral features

        l1 = (
            self.lateral_convs[0](
                c1
            )
        )

        l2 = (
            self.lateral_convs[1](
                c2
            )
        )

        l3 = (
            self.lateral_convs[2](
                c3
            )
        )

        # Top-down FPN

        l3 = l3 + (

            F.interpolate(

                p4,

                size=l3.shape[-2:],

                mode="bilinear",

                align_corners=False

            )

        )

        l2 = l2 + (

            F.interpolate(

                l3,

                size=l2.shape[-2:],

                mode="bilinear",

                align_corners=False

            )

        )

        l1 = l1 + (

            F.interpolate(

                l2,

                size=l1.shape[-2:],

                mode="bilinear",

                align_corners=False

            )

        )

        # FPN refinement

        f1 = self.fpn_convs[0](
            l1
        )

        f2 = self.fpn_convs[1](
            l2
        )

        f3 = self.fpn_convs[2](
            l3
        )

        f4 = self.fpn_convs[3](
            p4
        )

        target_size = (
            f1.shape[-2:]
        )

        f2 = F.interpolate(

            f2,

            size=target_size,

            mode="bilinear",

            align_corners=False

        )

        f3 = F.interpolate(

            f3,

            size=target_size,

            mode="bilinear",

            align_corners=False

        )

        f4 = F.interpolate(

            f4,

            size=target_size,

            mode="bilinear",

            align_corners=False

        )

        # Concatenate all levels

        x = torch.cat(

            [

                f1,
                f2,
                f3,
                f4

            ],

            dim=1

        )

        x = self.fpn_bottleneck(
            x
        )

        return self.classifier(
            x
        )
```

```python
decoder = UPerNetDecoder(

    in_channels=1280,

    decoder_channels=256,

    num_classes=2

)

print(
    "Decoder parameters:",
    sum(
        p.numel()
        for p in decoder.parameters()
    ) / 1e6,
    "M"
)
```

```python
class PrithviUPerNet(nn.Module):

    def __init__(
        self,
        encoder,
        decoder
    ):

        super().__init__()

        self.encoder = encoder

        self.decoder = decoder

    def forward(
        self,
        x
    ):

        features = (
            prithvi_features(
                self.encoder,
                x
            )
        )

        features = (
            tokens_to_feature_maps(
                features
            )
        )

        logits = self.decoder(
            features
        )

        logits = F.interpolate(

            logits,

            size=x.shape[-2:],

            mode="bilinear",

            align_corners=False

        )

        return logits
```

```python
class GSSDDataset(Dataset):

    def __init__(
        self,
        root="/content/GSSD/GSSD",
        split="test",
        augment=False
    ):

        self.image_dir = os.path.join(
            root,
            split,
            "image"
        )

        self.mask_dir = os.path.join(
            root,
            split,
            "mask"
        )

        self.augment = augment

        self.image_files = sorted(

            glob.glob(

                os.path.join(

                    self.image_dir,

                    "image*.tif"

                )

            ),

            key=lambda x: int(

                os.path.basename(x)

                .replace(
                    "image",
                    ""
                )

                .replace(
                    ".tif",
                    ""
                )

            )

        )

        self.mask_files = [

            os.path.join(

                self.mask_dir,

                "mask"
                +
                os.path.basename(x)
                .replace(
                    "image",
                    ""
                )

            )

            for x in self.image_files

        ]

        for img, mask in zip(

            self.image_files,

            self.mask_files

        ):

            if not os.path.exists(mask):

                raise FileNotFoundError(

                    f"Missing mask:\n"
                    f"{mask}"

                )

        print(

            f"{split.upper()} dataset loaded: "
            f"{len(self.image_files)} samples"

        )

    def __len__(self):

        return len(
            self.image_files
        )

    def __getitem__(self, idx):

        # --------------------------------
        # Read first 6 spectral bands
        # --------------------------------

        with rasterio.open(

            self.image_files[idx]

        ) as src:

            image = src.read(

                [
                    1,  # Blue
                    2,  # Green
                    3,  # Red
                    4,  # NIR
                    5,  # SWIR1
                    6   # SWIR2
                ]

            ).astype(
                np.float32
            )

        # --------------------------------
        # Read mask
        # --------------------------------

        with rasterio.open(

            self.mask_files[idx]

        ) as src:

            mask = src.read(
                1
            ).astype(
                np.float32
            )

        # --------------------------------
        # NumPy → Tensor
        # --------------------------------

        image = torch.from_numpy(
            image
        )

        mask = torch.from_numpy(
            mask
        )

        # --------------------------------
        # Resize image
        # --------------------------------

        image = F.interpolate(

            image.unsqueeze(0),

            size=(448, 448),

            mode="bilinear",

            align_corners=False

        ).squeeze(0)

        # --------------------------------
        # Resize mask
        # --------------------------------

        mask = F.interpolate(

            mask
            .unsqueeze(0)
            .unsqueeze(0),

            size=(448, 448),

            mode="nearest"

        )

        mask = (
            mask
            .squeeze(0)
            .squeeze(0)
        )

        # --------------------------------
        # Convert mask to binary
        # --------------------------------

        mask = (
            mask > 0.5
        ).long()

        return image, mask
```

```python
TP = 0
TN = 0
FP = 0
FN = 0

total_test_loss = 0.0

print(
    "Global metric counters initialized."
)
```

```python
net.eval()

with torch.no_grad():

    for step, (images, masks) in enumerate(
        test_loader
    ):

        images = images.to(
            device,
            non_blocking=True
        )

        masks = masks.to(
            device,
            non_blocking=True
        )

        with torch.autocast(

            device_type="cuda",

            dtype=torch.float16

        ):

            logits = net(
                images
            )

        preds = logits.argmax(
            dim=1
        )

        # Flatten

        pred = preds.reshape(
            -1
        )

        target = masks.reshape(
            -1
        )

        # Global confusion counts

        TP += (
            (
                (pred == 1)
                &
                (target == 1)
            )
            .sum()
            .item()
        )

        TN += (
            (
                (pred == 0)
                &
                (target == 0)
            )
            .sum()
            .item()
        )

        FP += (
            (
                (pred == 1)
                &
                (target == 0)
            )
            .sum()
            .item()
        )

        FN += (
            (
                (pred == 0)
                &
                (target == 1)
            )
            .sum()
            .item()
        )

        # Release batch tensors

        del images
        del masks
        del logits
        del preds
        del pred
        del target

        if (
            step + 1
        ) % 10 == 0:

            print(
                f"Processed "
                f"{step + 1}/"
                f"{len(test_loader)}"
            )

        if torch.cuda.is_available():

            torch.cuda.empty_cache()

print(
    "\n🔥 TEST INFERENCE COMPLETE."
)
```

```python
eps = 1e-7

precision = (
    TP
    /
    (
        TP + FP + eps
    )
)

recall = (
    TP
    /
    (
        TP + FN + eps
    )
)

f1 = (
    2
    *
    precision
    *
    recall
    /
    (
        precision
        +
        recall
        +
        eps
    )
)

iou_debris = (
    TP
    /
    (
        TP
        +
        FP
        +
        FN
        +
        eps
    )
)

iou_background = (
    TN
    /
    (
        TN
        +
        FP
        +
        FN
        +
        eps
    )
)

miou = (
    iou_background
    +
    iou_debris
) / 2.0
```
